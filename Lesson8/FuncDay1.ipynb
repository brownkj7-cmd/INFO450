```python
def celcius_to_fahrenheit(temperatures):
  new_temp=[]
  for temp in temperatures:
    fahreinheit=(temp*9/5)+32
    new_temp.append(fahreinheit)
  return new_temp
```

```python
celsius_temp=[0,1,2,3]
```

```python
celcius_to_fahrenheit(celsius_temp)
```

```python
product_prices=[10,15,20,25,30]
product_quantities=[100,200,300,400, 500]
def calculate_revenue(prices,quantities):
  """This fucntion takes prices and quantities and returns a  list of revenues"""
  revenues=[]
  for price,quantity in zip(prices,quantities):
    revenue=price*quantity
    revenues.append(revenue)
  return(revenues)
```
